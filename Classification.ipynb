```python
from collections import deque

def bfs():
 cpcity1 = 4
 cpcity2 = 3
 goal = 2


 queue = deque([((0,0),[])])
 visited = set()

 while queue:
  (x,y), path = queue.popleft()

  if(x,y) in visited:
    continue

  visited.add((x,y))
  path = path +[(x,y)]
  if x==goal:
      print("BFS Solution: ")
      for state in path:
        print(state)
      return

  next_states = [
          (cpcity1,y),(x,cpcity2),
          (0,y),(x,0),
          (min(cpcity1,x+y),max(0,x+y-cpcity1)),
          (max(0,x+y-cpcity2),min(cpcity2,x+y))
      ]

  for state in next_states:
      if state not in visited:
         queue.append((state,path))

bfs()
```

```python
def dfs():
    capacity1 = 4
    capacity2 = 3
    goal = 2

    stack = [((0, 0), [])]
    visited = set()

    while stack:
        (x, y), path = stack.pop()

        if (x, y) in visited:
            continue

        visited.add((x, y))
        path = path + [(x, y)]

        if x == goal:
            print("DFS Solution:")
            for state in path:
                print(state)
            return

        next_states = [
            (capacity1, y),              # Fill Jug 1
            (x, capacity2),              # Fill Jug 2
            (0, y),                      # Empty Jug 1
            (x, 0),                      # Empty Jug 2
            (min(capacity1, x + y),
             max(0, x + y - capacity1)), # Jug 2 -> Jug 1
            (max(0, x + y - capacity2),
             min(capacity2, x + y))       # Jug 1 -> Jug 2
        ]

        for state in next_states:
            if state not in visited:
                stack.append((state, path))

dfs()
```
